## Subplots and Legends

```python
days=[1,2,3,4,5]
orignal_price=[10,20,25,30,45]
predicted_price=[8,21,25,28,38]
```

```python
plt.plot(days,orignal_price,'b-o',label='orignal_price')
plt.plot(days,predicted_price,'r-o',label='predicted_price')
plt.legend()
plt.title('Plot with legends');
plt.xlabel('weekdays');
plt.ylabel('Price');
plt.savefig('example_image.png', dpi=300)
```

## Subplots

```python
# first subplot
plt.subplot(3,1,1)
plt.plot(days,orignal_price,'b-o',label='orignal_price')
plt.legend()
plt.title('Plot with legends');
plt.xlabel('weekdays');
plt.ylabel('Price');


#second sunplot
plt.subplot(3,1,2)
plt.plot(days,predicted_price,'r-o',label='predicted_price')
plt.legend()
plt.title('Plot with legends');
plt.xlabel('weekdays');
plt.ylabel('Price');

# diffrence between orignal and predicted price
price_diff=[o - p for o,p in zip(orignal_price,predicted_price)]
plt.subplot(3,1,3)
plt.bar(days,price_diff,color='green',label='Price diff')
plt.legend()
plt.title('Plot with legends');
plt.xlabel('weekdays');
plt.ylabel('Price');

plt.tight_layout()

```
